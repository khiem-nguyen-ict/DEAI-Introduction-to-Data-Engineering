First run the following code in Python:

# Explaination:
```
1. pd.merge performs a join between two dataframes, combining them into one dataframe.
2. on="country" — the country column is the join key. It must exist in both dataframes as preimary key. Rows are matched by comparing country values.
3. Default behavior is an inner join (similar to SQL database): 
    - Only rows where country exists in both dataframes survive. 
    - Countries present in one but not the other are dropped.
    - Different columns from both dataframes will be combined into the new dataframe.
```

````python
import pandas as pd

df_population = pd.DataFrame({
    "country": ["Finland", "Japan", "Bhutan", "Tuvalu"],
    "population": [5_620_000, 123_975_000, 790_000, 10_000]
})

df_gdp = pd.DataFrame({
    "country": ["Finland", "Japan", "Bhutan", "Tuvalu"],
    "gdp": [300, 4_200, 3.0, 0.06]
})
````

````python
df = pd.merge(df_population, df_gdp, on = "country")
df.head()
````
