````python
import pandas as pd

afghanistan_df = pd.read_csv("./countries/Afghanistan.csv")
albania_df = pd.read_csv("./countries/Albania.csv")
algeria_df = pd.read_csv("./countries/Algeria.csv")
angola_df = pd.read_csv("./countries/Angola.csv")

df = pd.concat([afghanistan_df, albania_df, algeria_df, angola_df], ignore_index=True)
df.head()
````

````python
# Lamda function for each data group (by Entity, DataFrameGroupBy)
def analyze_entity(dg):
    min_year = dg["Year"].min()
    max_year = dg["Year"].max()
    min_gdp = dg.loc[dg["Year"].idxmin(), "GDP"] / 10**9
    max_gdp = dg.loc[dg["Year"].idxmax(), "GDP"] / 10**9
    avg = dg["GDP"].mean() / 10**9
    n = max_year - min_year
    cagr = round(((max_gdp / min_gdp) ** (1 / n) - 1) * 100, 2)
    # Create a new row with data
    return pd.Series({
        "Min Year": min_year,
        "Max Year": max_year,
        "Min GDP (B$)": round(min_gdp, 2),
        "Max GDP (B$)": round(max_gdp, 2),
        "Avg GDP (B$)": round(avg, 2),
        "CAGR %": cagr,
    })

# DataFrameGroupBy is a pandas class — the object returned by df.groupby(...).
print(type(df.groupby("Entity"))) # --> DataFrameGroupBy
# For each group, call analyze_entity to calculate the create new series (row) data for them
results = df.groupby("Entity")[["Year", "GDP"]].apply(analyze_entity).reset_index()
results
````
