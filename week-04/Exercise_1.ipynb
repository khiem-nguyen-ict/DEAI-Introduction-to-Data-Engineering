# Exercise 1 - Count the choices with value_counts

Read `week-04/materials/2026-DEAI-intro-data-engineering-people-preferences-dataset.csv` with `sep=";"`.

A) Run `value_counts()` on Favorite Food, Favorite Pet, and Music. For each column, what is the most common choice and how many people chose it?

B) Run `value_counts(normalize=True)` on Favorite Season. What share of people chose summer?

```python
import pandas as pd

df = pd.read_csv("./materials/2026-DEAI-intro-data-engineering-people-preferences-dataset.csv", sep=";")

# a) Run value_counts on Favorite Food, Favorite Pet, and Music

print("Favorite Foods: ", df["Favorite Food"].value_counts())
print("\r\nFavorite Pets: ", df["Favorite Pet"].value_counts())
print("\r\nMusic: ", df["Music"].value_counts())

# According to the result, we have the common choice and number of people for each is:
# Favorite Food → Pizza, 22 people
# Favorite Pet → Dog, 26 people
# Music → Pop, 14 people

# b) Run value_counts(normalize=True)
df["Favorite Season"].value_counts(normalize=True)

# The share if people who choosed Summer is 0.5532 (≈55.3%), i.e. 26 of 47 people.
```
