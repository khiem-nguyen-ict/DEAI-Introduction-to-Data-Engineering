# Exercise 2 - value_counts inside a group

Use the survey from Exercise 1. First run `value_counts(normalize=True)` on Favorite Pet.

A) Keep only the rows where Favorite Pet is dog. Run `value_counts(normalize=True)` on Dream Holiday Destination. Which category changes most which least?  

B) Do the same analysis for Favorite Pet equal to cat. Which destination shows up for cats and did not show up for dogs?

C) Keep only the rows where Favorite Season is summer, and run `value_counts(normalize=True)` on Favorite Pet. Then do the same for winter. In both seasons, which pet is the most common?

```python
import pandas as pd

df = pd.read_csv("./materials/2026-DEAI-intro-data-engineering-people-preferences-dataset.csv", sep=";")

# a) 

print("Which category changes most which least", df.loc[df["Favorite Pet"] == "Dog", "Dream Holiday Destination"].value_counts(normalize=True))

# The most category changed is "Hawaii", 0.384615, the least is SanTorini, 0.115385

# b) 
df.loc[df["Favorite Pet"] == "Cat", "Dream Holiday Destination"].value_counts(normalize=True)

d_dog = df.loc[df["Favorite Pet"] == "Dog", "Dream Holiday Destination"].unique()
d_cat = df.loc[df["Favorite Pet"] == "Cat", "Dream Holiday Destination"].unique()

d_result = set(d_cat) - set(d_dog)

print("Which destination shows up for cats and did not show up for dogs:", d_result)

# c) 

print("In Summer, Common Pet is: ", df.loc[df["Favorite Season"] == "Summer", "Favorite Pet"].value_counts(normalize=True))
print("In Winter, Common Pet is: ", df.loc[df["Favorite Season"] == "Winter", "Favorite Pet"].value_counts(normalize=True))

# So, in both seasons, Dog is still the most common.

```
