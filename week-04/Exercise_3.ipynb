````python
from itertools import permutations
import pandas as pd

df = pd.read_csv("./materials/2026-DEAI-intro-data-engineering-people-preferences-dataset.csv", sep=";")

columns = [
    "Your favorite color",
    "Dream Holiday Destination",
    "Favorite Pet",
    "Favorite Season",
    "Music",
    "Favorite Food",
]

trios = list(permutations(columns, 3))

frames = []
for cols in trios[:3]:
    v = df.loc[:, cols].value_counts()
    v = v[v > 1]
    v = v.to_frame(name="count")
    v.insert(0, "trio", ", ".join(cols))
    frames.append(v)

# Concat the three permutations into one long table
result = pd.concat(frames, axis=0, ignore_index=True)

with pd.option_context("display.max_colwidth", None):
    print(result)
````
