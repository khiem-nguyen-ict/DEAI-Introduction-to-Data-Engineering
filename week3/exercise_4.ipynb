````python
import pandas as pd

df1 = pd.DataFrame({
    "country": ["Belize", "Liechtenstein", "Bhutan"],
    "population": [417_000, 40_000, 790_000]
})

df2 = pd.DataFrame({
    "country": ["Suriname", "São Tomé and Príncipe", "Tuvalu"],
    "population": [630_000, 240_000, 10_000]
})

# Stack them first
df = pd.concat([df1, df2], ignore_index=True)
df

# To the column "population":

print(f"1. Mean = {df["population"].mean()}: average, by adding up all values, divide by count.")
print(f"2. Sum = {df["population"].sum()}: The sum value of population (total)")
print(f"3. Min = {df["population"].min()}: The minimum value of population")
print(f"4. Max = {df["population"].max()}: The maximum value of population")
print(f"5. Count = {df["population"].count()}: The number of items")
print(f"6. Median = {df["population"].median()}: The middle value, by sorting the data, pick the middle value.")
print(f"7. Standard Deviation (std) = {df["population"].std()}: It measures how spread out population data is around the mean vaue.")
print(f"8. Variance (var) = {df["population"].var()}: standard deviation squared, = std^2")
````
