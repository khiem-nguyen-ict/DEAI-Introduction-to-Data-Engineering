### Filtering

Select only lines where height is over 1.5 m

```python
df_characters[df_characters["height"] > 1.5].head(3)
```

Under the hood pandas produces a boolean array and that is used to make selection

```python
df_characters["height"] > 1.5
```

Filtering with multiple conditions when both conditions must be true, we use **&**. Syntax is the following: **df_characters[ (condition 1) & (condintion 2)]** Note that we use & instead of **and**.

```python
df_characters[ ( df_characters["species"] == "Human") & ( df_characters["height"] > 2.0 ) ]
```

**Pandas logical operators are:**

& = bitwise AND operator

| = bitwise OR operator

~ = bitwise NOT operator

We can combine the operators. Here we choose characters whose eye_color is **not** Blue nor Brown or whose hair color is white

```python
df_characters[ (  ~ df_characters["eye_color"].isin(["Blue","Brown"]) ) | ( df_characters["hair_color"] == "White" ) ].head(5)
```

**Let's take a subset of the data**

**Question 1**: What is happening in the code below? Which lines are selected and why?

```python
df_droids[[False,True,True,True]]
# WHAT IS HAPPENING:
# Direct bracket indexing with a list of booleans filters the DataFrame rows by positional order.
#
# WHICH LINES ARE SELECTED:
# Rows at positional index 1, 2, and 3 (the 2nd, 3rd, and 4th rows of the DataFrame).
#
# WHY:
# Pandas maps each boolean value 1-to-1 to the DataFrame's rows in sequential order:
# - Position 0 (1st row) -> False (Excluded)
# - Position 1 (2nd row) -> True  (Included)
# - Position 2 (3rd row) -> True  (Included)
# - Position 3 (4th row) -> True  (Included)
#
# NOTE: The boolean list length must EXACTLY match the total number of rows in df_droids,
# otherwise Pandas will raise a ValueError.
```

**Question 2**: Why does this code produce the similar output?

```python
df_droids[df_droids["name"].str.len() == 5]

# WHAT IS HAPPENING:
# Filters df_droids to return rows where the string length of the 'name' column is exactly 5 characters.
#
# WHY IT PRODUCES A SIMILAR OUTPUT:
# The similar result is purely coincidental (a "lucky case") based on the current dataset:
# the last 3 rows happen to have 5-letter names (e.g., 'C-3PO', 'R2-D2', or 'L3-37' depending on data), 
# while the 1st row does not ('BB-8' has only 4 characters).
#
# CRITICAL DIFFERENCE:
# - Previous code: Hardcoded row positions (positional selection via a boolean list).
# - This code: Dynamic conditional filtering based on data values.
# If the 'name' values change (becoming shorter or longer than 5 characters), 
# the returned rows will change accordingly.
```
